```python
# 5. Visualize Results
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os

def show_image(path, title):
    if os.path.exists(path):
        img = mpimg.imread(path)
        plt.figure(figsize=(10, 6))
        plt.imshow(img)
        plt.title(title)
        plt.axis('off')
        plt.show()
    else:
        print(f"Image not found: {path}")

results_dir = 'results'

# Show key graphs
show_image(os.path.join(results_dir, 'loss_curve.png'), 'Training Loss')
show_image(os.path.join(results_dir, 'scatter_pred.png'), 'Prediction Accuracy')
show_image(os.path.join(results_dir, 'heatmap_comparison.png'), 'Correlation Matrix Heatmap')
show_image(os.path.join(results_dir, 'time_series_pair.png'), 'Time Series Forecast')
```
